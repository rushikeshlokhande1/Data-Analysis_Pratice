```python
df = pd.read_csv(
    "zomato.csv",
    engine="python",
    on_bad_lines="skip"
)
```

```python
df.info()
```

```python
drop_columns=['url', 'address', 'votes',
       'phone', 'location', 'rest_type',  'cuisines',
       'listed_in(type)', 'listed_in(city)']
```

```python
df=df.drop(columns=drop_columns)
```

```python
df.head()
```

```python
new_column=[]
for i in df.columns:
  new_column.append(i.upper())

```

```python
df.columns=new_column
```

```python
df.head()
```

```python
df=df.drop(columns=['REVIEWS_LIST','MENU_ITEM'])
```

```python
df.head()
```

```python
df['ONLINE_ORDER']=df['ONLINE_ORDER'].apply(lambda x:1 if x=='Yes' else 0)
```

```python
df.head()
```

```python
df.duplicated().sum()
```

```python
df=df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df['RATE']=df['RATE'].replace("/5","")
```

```python
df.isnull().sum()
```

```python
df['APPROX_COST(FOR TWO PEOPLE)'] = df['APPROX_COST(FOR TWO PEOPLE)'].str.replace(',', '')
```

```python
df['APPROX_COST(FOR TWO PEOPLE)'] = pd.to_numeric(
    df['APPROX_COST(FOR TWO PEOPLE)'],
    errors='coerce'
)
```

```python
df['APPROX_COST(FOR TWO PEOPLE)']=df['APPROX_COST(FOR TWO PEOPLE)'].fillna(df['APPROX_COST(FOR TWO PEOPLE)'].mean())
```

```python
df['RATE'] = df['RATE'].str.replace('/5', '', regex=False)
```

```python
df
```

```python
df['BOOK_TABLE']=df['BOOK_TABLE'].apply(lambda x:1 if x=='Yes' else 0)
```

```python
df
```

```python
df=df.dropna()
```

```python
df['RATE'] = pd.to_numeric(df['RATE'], errors='coerce')
```

```python
df['RATE']=df['RATE'].astype(float)
```

```python
df.head(3)
```

```python
df=df.reset_index(drop=True)
```

```python
df
```

```python
df.to_csv('zomato_cleaned.csv')
```
