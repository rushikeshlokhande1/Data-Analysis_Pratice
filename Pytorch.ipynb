```python
import torch

# Training data
x = torch.tensor([1., 2., 3., 4., 5.])
y = torch.tensor([2., 3., 4., 5., 6.])

# Parameters (weight and bias)
w = torch.tensor(0.0, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)

# Hyperparameters
lr = 0.1
epochs = 10

# Training loop
for epoch in range(epochs):
    # Forward pass
    y_pred = w * x + b

    # Compute loss (Mean Squared Error)
    loss = ((y_pred - y) ** 2).mean()

    # Backward pass
    loss.backward()

    # Update parameters
    with torch.no_grad():
        w -= lr * w.grad
        b -= lr * b.grad

    # Reset gradients
    w.grad.zero_()
    b.grad.zero_()

    # Print progress every 100 epochs
    if (epoch + 1) % 100 == 0:
        print(
            f"Epoch {epoch+1}: "
            f"w = {w.item():.4f}, "
            f"b = {b.item():.4f}, "
            f"loss = {loss.item():.6f}"
        )

print("\nTraining Complete!")
print(f"Learned Weight (w): {w.item():.4f}")
print(f"Learned Bias (b): {b.item():.4f}")
import torch

# Training data
x = torch.tensor([1., 2., 3., 4., 5.])
y = torch.tensor([2., 3., 4., 5., 6.])

# Parameters (weight and bias)
w = torch.tensor(0.0, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)

# Hyperparameters
lr = 0.01
epochs = 1000

# Training loop
for epoch in range(epochs):
    # Forward pass
    y_pred = w * x + b

    # Compute loss (Mean Squared Error)
    loss = ((y_pred - y) ** 2).mean()

    # Backward pass
    loss.backward()

    # Update parameters
    with torch.no_grad():
        w -= lr * w.grad
        b -= lr * b.grad

    # Reset gradients
    w.grad.zero_()
    b.grad.zero_()

    # Print progress every 100 epochs
    if (epoch + 1) % 100 == 0:
        print(
            f"Epoch {epoch+1}: "
            f"w = {w.item():.4f}, "
            f"b = {b.item():.4f}, "
            f"loss = {loss.item():.6f}"
        )

print("\nTraining Complete!")
print(f"Learned Weight (w): {w.item():.4f}")
print(f"Learned Bias (b): {b.item():.4f}")

# Test the model
test_x = torch.tensor([6.0])
predicted_y = w * test_x + b
print(f"\nFor x = {test_x.item()}, Predicted y = {predicted_y.item():.4f}")
```
